```python
import pandas as pd
import json
```

```python
xl_path = '../data/xl/dados_metas_atualizados_v2.xlsx'
```

```python
df = pd.read_excel(xl_path, engine='openpyxl', sheet_name='acoes_estrategicas')
```

```python
df.head()
```

```python
renomear = {
    'Número da Meta': 'numero_meta',
    'Número da Ação Estratégica' : 'numero_acao' ,
    'Descrição da Ação Estratégica' : 'desc_acao',
    'Secretaria' : 'orgao_acao'
}
```

```python
df.rename(columns=renomear, inplace=True)
```

```python
df = df[renomear.values()].copy()
```

```python
df.head()
```

```python
renomear_orgaos = {
    'SP REGULA' : 'SPRegula',
    'SPTRANS' : 'SPTrans',
    'SP TRANS' : 'SPTrans',
    'SP OBRAS' : 'SPObras',
    'SGM/SEDP' : 'SEDP',
    'SP URBANISMO' : 'SPUrbanismo',
    'SGM/SEPE' : 'SEPE',
    'SMDHC/SESANA' : 'SESANA',
    'SUBPREFEITURAS' : 'SMSUB'

}

for atual, novo in renomear_orgaos.items():

    df['orgao_acao'] = df['orgao_acao'].str.replace(atual, novo)
```

```python
acoes_parsed = []

for i, row in df.iterrows():

    orgaos = [secretaria.strip() for secretaria in row['orgao_acao'].split(';')
              if secretaria.strip()]

    acao = {
        'numero_meta': row['numero_meta'],
        'numero_acao': row['numero_acao'],
        'desc_acao': row['desc_acao'],
        'orgao_acao': orgaos
    }
    acoes_parsed.append(acao)
```

```python
with open('../data/acoes_estrategicas.json', 'w', encoding='utf-8') as f:
    json.dump(acoes_parsed, f, indent=4, ensure_ascii=False)
```
