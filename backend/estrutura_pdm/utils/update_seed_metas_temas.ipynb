## Existem temas novos

Como, para além das diferenças de grafias, parecem ter surgido temas novos, vou atualizar os temas em si.

```python
df_eixos = df[['eixo', 'tema']].drop_duplicates().reset_index(drop=True)
```

```python
df_eixos.head()
```

```python
def get_dados_eixo(nome_eixo):
    for eixo in dados_eixos:
        if eixo['nome'] == nome_eixo:
            return eixo
    raise ValueError(f'Eixo {nome_eixo} não encontrado')
```

```python
dados_temas_eixos = {}
new_dados_eixos = []

eixos = df_eixos['eixo'].unique()
for eixo in eixos:
    
    dados_eixo = get_dados_eixo(eixo)
    dados_eixo['temas'] = []
    temas_eixo = df_eixos[df_eixos['eixo']==eixo]['tema']

    for tema in temas_eixo:
        dados_eixo['temas'].append({
            'nome' : tema,
            'descricao' : 'XXXX'
        })

    new_dados_eixos.append(dados_eixo)


```

```python
new_dados_eixos[1]
```

```python
with open(path_dados_eixos, 'w') as f:
    json.dump(new_dados_eixos, f, indent=4, ensure_ascii=False)
```

Agora podemos criar o arquivo de metas e temas com segurança (pois os temas dos eixos foram atualizados com base na planilha)

```python
df.head()
```

```python
df_metas = df[['meta_numero', 'tema']].drop_duplicates().reset_index(drop=True)
```

```python
df_metas.shape
```

```python
df_metas['meta_numero'].duplicated().any()
```

```python
metas_temas = {}
for i, row in df_metas.iterrows():
    meta = row['meta_numero']
    if meta in metas_temas:
        raise ValueError(f'Meta {meta} duplicada')

    metas_temas[meta] = row['tema'].strip()
```

```python
metas_temas
```

```python
with open('../data/metas_tema.json', 'w') as f:
    json.dump(metas_temas, f, indent=4, ensure_ascii=False)
```
