```python
import pandas as pd
import json
```

```python
xl_path = '../data/xl/dados_metas_atualizados_v2.xlsx'
```

```python
df = pd.read_excel(xl_path, engine='openpyxl', sheet_name='Temas, Planos e ODS')
```

```python
df.head()
```

```python
renomear = {
    'Número da Meta' : 'numero_meta',
    'ODS' : 'ods'
}
```

```python
df.rename(columns=renomear, inplace=True)
```

```python
df = df[renomear.values()].copy()
```

```python
df.head()
```

```python
df['ods'].isnull().any()
```

```python
df['numero_meta'].isnull().any()
```

```python
df['numero_meta'].duplicated().any()
```

```python
parsed_data = {}

for i, row in df.iterrows():

    numero_meta = int(row['numero_meta'])

    ods = str(row['ods']).split(';')

    ods = [int(x.strip()) for x in ods]

    parsed_data[numero_meta] = ods
```

```python
parsed_data
```

```python
with open('../data/metas_ods.json', 'w') as f:
    json.dump(parsed_data, f, indent=4, ensure_ascii=False)
```
